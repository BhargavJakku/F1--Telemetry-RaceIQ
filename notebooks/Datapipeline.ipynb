# F1 Telemetry Datapipeline

```python
# Defining Constraints 
Early_season_races = ['Bahrain','Saudi Arabia','Australia']
Late_season_races = ['Monza','Japan','Las Vegas']
Year = 2024
```

```python
#Import Libs
import fastf1
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Using cache data
fastf1.Cache.enable_cache('../f1_Cache')
print('Cache loaded Successfully')
```

```python
# Functions
def pull_race_data(year,race_name):
    session = fastf1.get_session(year,race_name,'R')
    session.load()
    laps = session.laps.copy()
    laps['Race'] = race_name
    laps['Season'] = year
    return laps

all_laps = []

for race in Early_season_races + Late_season_races:
    print(f'Pulling Race:{race}')
    laps = pull_race_data(Year,race)
    all_laps.append(laps)
    print(f'Pulling Done:{len(laps)}')

raw_df = pd.concat(all_laps,ignore_index=True)
print(f'Total laps loaded:{len(raw_df)}')
    
```

```python
# function to clean laps 
def clean_laps(df):
    #keep only accurate laps 
    df = df[df['IsAccurate']==True].copy()
    #Change LapTime in seconds
    df['LapTimeSeconds'] = df['LapTime'].dt.total_seconds()
    #Drop Null values
    df = df.dropna(subset=['LapTimeSeconds','Compound','TyreLife'])
    #drop outliers
    df = df[df['LapTimeSeconds'].between(70,130)]

    return df

clean_df = clean_laps(raw_df)
print(f'Rows before cleaning:{len(raw_df)}, Rows after cleaning:{len(clean_df)}')
print(f'Rows deleted:{len(raw_df)-len(clean_df)}')
```

```python
# Feature Engineering
def race_feature_eng(df):

    # Encoding Tyre Compound
    compound_map = {
        'SOFT':0,
        'MEDIUM':1,
        'HARD':2,
        'INTERMEDIATE':3,
        'WET':4
    }
    df['compound_encoded'] = df['Compound'].map(compound_map)

    #Normalize Tyrelife like with each stint
    df['tyre_life_norm'] = df.groupby(['Race','Driver','Stint'])['TyreLife'].transform(
    lambda x: (x - x.min()) / (x.max() - x.min() + 1e-5))

    #Normalize Lapnumber 
    df['lapnumber_norm'] = df.groupby('Race')['LapNumber'].transform(lambda x: x/x.max())

    df['StintNumber'] = df['Stint'].fillna(1).astype(int)

    return df

featured_df = race_feature_eng(clean_df)
```

```python
featured_df
```

```python
#Finalizing Features
Features = ['compound_encoded','TyreLife','tyre_life_norm','LapNumber','lapnumber_norm','StintNumber']
Target = 'LapTimeSeconds'

final_df = featured_df[Features + [Target, 'Race','Driver']]
print(f'Final dataset shape:{final_df.shape}')
print(f'\n Feature Summary:')
print(final_df[Features].describe())
```

```python
final_df
```

```python
# Data Spilting 
dataset_a = final_df[final_df['Race'].isin(Early_season_races)].copy()
dataset_b = final_df[final_df['Race'].isin(Late_season_races)].copy()


print(f'Dataset_a :{len(dataset_a)}')
print(f'Dataset_b :{len(dataset_b)}')
dataset_a.to_csv('../data/spilts/dataset_a.csv',index=False)
dataset_b.to_csv('../data/spilts/dataset_b.csv',index=False)

print('Data Saved successfully')
```

```python
# Validating Data 
print("Dataset A races:", dataset_a['Race'].unique())
print("Dataset B races:", dataset_b['Race'].unique())
print("\nDataset A lap time range:", dataset_a['LapTimeSeconds'].min(), 
      "—", dataset_a['LapTimeSeconds'].max())
print("Dataset B lap time range:", dataset_b['LapTimeSeconds'].min(), 
      "—", dataset_b['LapTimeSeconds'].max())
print("\nCompound distribution in A:")
print(dataset_a['compound_encoded'].value_counts())
print("\nCompound distribution in B:")
print(dataset_b['compound_encoded'].value_counts())
```
